```python
# =========================================
import os
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

from google.colab import files

print("Upload training_setA folder (zipped)")
uploaded_A = files.upload()  # Upload your training_setA.zip
print("Upload training_setB folder (zipped)")
uploaded_B = files.upload()  # Upload your training_setB.zip

# Unzip
import zipfile

with zipfile.ZipFile("training_setA.zip", 'r') as zip_ref:
    zip_ref.extractall("training_setA")

with zipfile.ZipFile("training_setB.zip", 'r') as zip_ref:
    zip_ref.extractall("training_setB")

# =========================================
# STEP 3: Read all PSV files into DataFrames
# =========================================
def load_psv_folder(folder_path):
    all_files = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.endswith(".psv")]
    dfs = [pd.read_csv(file, sep="|") for file in all_files]
    return pd.concat(dfs, ignore_index=True)

trainA = load_psv_folder("training_setA")
trainB = load_psv_folder("training_setB")

print("Training A shape:", trainA.shape)
print("Training B shape:", trainB.shape)
```
